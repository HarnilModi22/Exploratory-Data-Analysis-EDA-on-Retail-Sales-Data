# Exploratory Data Analysis (EDA) on Retail Sales Data

Retail Sales Data Analysis is an Exploratory Data Analysis (EDA) project used to analyze sales and product-related data to identify patterns, trends, and insights. Pandas and NumPy are used for data handling, while Seaborn and Matplotlib are used for visualization. The project helps businesses make data-driven decisions using statistical and graphical analysis.

# Import Libraries

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

# Load Dataset

```python
df = pd.read_csv("menu.csv")
```

```python
df.head()
```

```python
df.info()
```

# Checking Missing Values

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)

df.shape
```

# Descriptive Statistics

```python
df.describe()
```

# Mean

```python
df.mean(numeric_only=True)
```

# Median

```python
df.median(numeric_only=True)
```

# Mode

```python
# Mode for numeric columns only

df.select_dtypes(include=np.number).mode().head(1)
```

# Standard Deviation

```python
df.std(numeric_only=True)
```

# Data Visualization

```python
plt.figure(figsize=(12,8))

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")

plt.show()
```

# Bar Chart

```python
top_categories = df['Category'].value_counts().head(10)

plt.figure(figsize=(10,6))

top_categories.plot(kind='bar')

plt.title("Top Categories")

plt.xlabel("Category")

plt.ylabel("Count")

plt.show()
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

df[numeric_cols].mean().plot(kind='line', figsize=(10,6))

plt.title("Average Nutritional Values")

plt.ylabel("Value")

plt.show()
```

# Customer and Product Analysis

```python

top_products = df[['Item', 'Calories']].sort_values(
    by='Calories',
    ascending=False
).head(10)

top_products
```

# Visualization of High Calorie Products

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x='Calories',
    y='Item',
    data=top_products
)

plt.title("Top 10 High Calorie Products")

plt.show()
```

# Result

The retail sales dataset was successfully analyzed using EDA techniques. Important insights were identified through descriptive statistics and visualizations, helping understand product trends, nutritional patterns, and category distributions.
